```python
#Importing pandas for data manipulation and numpy for vector operations
#Importing Standard Scalar for normalisation 
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
```

```python
#Importing uncleaned laptop Price dataset: https://www.kaggle.com/datasets/ehtishamsadiq/uncleaned-laptop-price-dataset

df = pd.read_csv("laptopData.csv")
```

```python
#Dispalying the first 5 records of the dataframe
df.head()
```

```python
#Displaying the infomation of the dataframe: includes number of entries, number of columns 
#as well as datatypes and other properties of attributes
df.info()
```

```python
#The dataframe consists of numeric as well as categorical data, describing all the data
df.describe(include='all')
```

```python
#Displaying the number of null values in each attribute to pre-process data
df.isnull().sum()
```

```python
#Data Cleaning: Dropping the unnamed column from the dataframe
df = df.drop(columns=['Unnamed: 0'])
```

```python
#The Inches attribute is an object type hence converting it to numeric using pd.to_numeric
#Inches column has '?' values so replacing those with NaN
#Replacing all NaN values in Inches column with the mean of the column using fillna
df['Inches'] = df['Inches'].replace('?', np.nan)
df['Inches'] = pd.to_numeric(df['Inches'], errors='coerce')
df['Inches'] = df['Inches'].fillna(df['Inches'].mean())

#The Weight column is of object datatype in the form "25 kg"
#Hence we replace the kg with empty string '' using str.replace
#We remove '?' values with NaN using replace
#Replace all NaN values with mean of the Weight column
df['Weight'] = df['Weight'].replace('?', np.nan)
df['Weight'] = df['Weight'].str.replace('kg', '', regex=False)
df['Weight'] = pd.to_numeric(df['Weight'], errors='coerce')
df['Weight'] = df['Weight'].fillna(df['Weight'].mean())

#Replacing all NaN values in Price column with its mean
df['Price'] = df['Price'].fillna(df['Price'].mean())
```

```python
#num_cols = ['Unnamed: 0', 'Price', 'Inches', 'Weight']
#for col in num_cols:
 #   df[col] = df[col].fillna(df[col].mean())
```

```python
#There esists categorical data in the dataset
#Replacing the NaN mode (highest frequency values of that columns hence mode()[0] since we replace with single value not Series)
cat_cols = ['Company', 'TypeName', 'ScreenResolution', 'Cpu', 'Ram', 'Memory', 'Gpu', 'OpSys']
for col in cat_cols:
    df[col] = df[col].fillna(df[col].mode()[0])
```

```python
#Chceking if the data cleaning removed NaN values in all attributes
df.isna().sum()
```

```python
#Performing Aggregation on the numeric columns to find mean, median, minimum, maximum and standard deviation
df[['Price', 'Inches', 'Weight']].agg(['mean', 'median', 'min', 'max', 'std'])
```

```python
#Finding the mean price of laptops of each company using groupby
df.groupby('Company')['Price'].mean()
```

```python
#Filtering out which Dell laptops are less than 20,000
df[(df['Company'] == 'Dell') & (df['Price'] < 20000)]
```

```python
#Normalising Weight attribute using standard Scalar
scaler = StandardScaler()
df['Weight_std'] = scaler.fit_transform(df[['Weight']])
```

```python
#Values less than mean of Weight are negative
#values equal to mean are 0
#Values greater than mean values are positive
df['Weight_std']
```

```python
#Finding the weight of least weight laptops of all companies
df.groupby('Company')['Weight_std'].min()
```

```python
#Finding the light weight lqaptops count of each company
df[df['Weight_std'] < 0].groupby('Company').size()
```

```python
#Finding the heavy weight lqaptops count of each company
df[df['Weight_std'] > 0].groupby('Company').size()
```
