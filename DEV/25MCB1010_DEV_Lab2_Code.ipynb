Web Scraping using BeautifulSoup

```python
import requests
from bs4 import BeautifulSoup
```

```python
def fetch_html(url):
    response = requests.get(url)
    if response.status_code == 200:
        return response.text
    else:
        raise Exception(f"Failed to fetch page: {response.status_code}")
```

```python
def parse_html(html_content):
    return BeautifulSoup(html_content, "html.parser")
```

```python
def extract_quotes(soup):
    quotes_data = []
    quotes = soup.find_all("div", class_="quote")
    for q in quotes:
        text = q.find("span", class_="text").get_text()
        author = q.find("small", class_="author").get_text()
        quotes_data.append({"quote": text, "author": author})
    return quotes_data
```

```python
def extract_tags(soup):
    tags_list = []
    quotes = soup.find_all("div", class_="quote")
    for q in quotes:
        tags = [tag.get_text() for tag in q.find_all("a", class_="tag")]
        tags_list.append(tags)
    return tags_list
```

```python
def scrape_page(url):
    html = fetch_html(url)
    soup = parse_html(html)
    quotes = extract_quotes(soup)
    tags = extract_tags(soup)
    return {"quotes": quotes, "tags": tags}
```

```python
def scrape_multiple_pages(base_url, num_pages=1):
    all_data = []
    for page in range(1, num_pages + 1):
        url = f"{base_url}/page/{page}/"
        try:
            page_data = scrape_page(url)
            all_data.extend(page_data["quotes"])
        except Exception as e:
            print(f"Error scraping {url}: {e}")
    return all_data
```

```python
if __name__ == "__main__":
    base_url = "http://quotes.toscrape.com"
    html = fetch_html(base_url)
    soup = parse_html(html)
```

```python
quotes = extract_quotes(soup)

print("Quotes from first page:")
for q in quotes:
  print(f"{q['quote']} — {q['author']}")
```

```python
tags = extract_tags(soup)

print("\nTags from first page:")
for t in tags:
  print(t)
```

```python
print("\nScraping multiple pages...")
data = scrape_multiple_pages(base_url, num_pages=2)
for item in data:
    print(f"{item['quote']} — {item['author']}")
```
